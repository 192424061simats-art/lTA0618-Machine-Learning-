```python
import pandas as pd

data = pd.DataFrame([
    ['Sunny', 'Warm', 'Normal', 'Yes'],
    ['Sunny', 'Warm', 'High', 'Yes'],
    ['Rainy', 'Cold', 'High', 'No'],
    ['Sunny', 'Warm', 'High', 'Yes']
], columns=['Sky', 'Temp', 'Humidity', 'Enjoy'])

S = ['0', '0', '0']
G = [['?', '?', '?']]

for _, row in data.iterrows():
    x = list(row[:-1])

    if row['Enjoy'] == 'Yes':
        for i in range(3):
            if S[i] == '0':
                S[i] = x[i]
            elif S[i] != x[i]:
                S[i] = '?'
    else:
        new_G = []
        for g in G:
            for i in range(3):
                if g[i] == '?' and S[i] != '?' and S[i] != x[i]:
                    temp = g.copy()
                    temp[i] = S[i]
                    new_G.append(temp)
        G = new_G

print("Specific Boundary:", S)
print("General Boundary:", G)
```
